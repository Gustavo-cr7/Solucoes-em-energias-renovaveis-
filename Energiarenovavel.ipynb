```python
import streamlit as st
import pandas as pd
import random

print("=" * 60)
print("SPACE ENERGY MONITOR")
print("Global Solution - Energias Renováveis e Sustentáveis")
print("=" * 60)

modulos = [
    "Painel Solar",
    "Bateria Central",
    "Comunicação",
    "Controle Térmico",
    "Suporte de Vida",
    "Propulsão"
]

dados = []

for modulo in modulos:
    temperatura = random.randint(15, 50)
    energia = random.randint(10, 100)
    geracao = random.randint(20, 100)
    consumo = random.randint(20, 100)
    status = random.choice(["Online", "Offline"])
    comunicacao = random.choice(["Estável", "Instável"])

    alerta = "Sistema normal"

    if temperatura > 40:
        alerta = "Temperatura crítica"

    elif energia < 25:
        alerta = "Energia baixa"

    elif status == "Offline":
        alerta = "Módulo offline"

    elif comunicacao == "Instável":
        alerta = "Falha de comunicação"

    dados.append({
        "Módulo": modulo,
        "Temperatura": temperatura,
        "Energia (%)": energia,
        "Geração (kW)": geracao,
        "Consumo (kW)": consumo,
        "Comunicação": comunicacao,
        "Status": status,
        "Alerta": alerta
    })

df = pd.DataFrame(dados)

print("\nDADOS DOS MÓDULOS:\n")
print(df)

print("\n" + "=" * 60)
print("ALERTAS DA MISSÃO")
print("=" * 60)

for _, linha in df.iterrows():

    if linha["Alerta"] != "Sistema normal":
        print(f"{linha['Módulo']}: {linha['Alerta']}")

print("\nSistema finalizado.")
```
