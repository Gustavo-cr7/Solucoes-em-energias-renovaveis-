```python
import streamlit as st
import pandas as pd
import random

print("=" * 60)
print("SPACE ENERGY MONITOR")
print("Global Solution - Energias Renováveis e Sustentáveis")
print("=" * 60)

modulos = [
    "Painel Solar",
    "Bateria Central",
    "Comunicação",
    "Controle Térmico",
    "Suporte de Vida",
    "Propulsão"
]

dados = []

for modulo in modulos:
    temperatura = random.randint(15, 50)
    energia = random.randint(10, 100)
    geracao = random.randint(20, 100)
    consumo = random.randint(20, 100)
    status = random.choice(["Online", "Offline"])
    comunicacao = random.choice(["Estável", "Instável"])

    alerta = "Sistema normal"

    if temperatura > 40:
        alerta = "Temperatura crítica"

    elif energia < 25:
        alerta = "Energia baixa"

    elif status == "Offline":
        alerta = "Módulo offline"

    elif comunicacao == "Instável":
        alerta = "Falha de comunicação"

    dados.append({
        "Módulo": modulo,
        "Temperatura": temperatura,
        "Energia (%)": energia,
        "Geração (kW)": geracao,
        "Consumo (kW)": consumo,
        "Comunicação": comunicacao,
        "Status": status,
        "Alerta": alerta
    })

df = pd.DataFrame(dados)

print("\nDADOS DOS MÓDULOS:\n")
print(df)

print("\n" + "=" * 60)
print("ALERTAS DA MISSÃO")
print("=" * 60)

for _, linha in df.iterrows():

    if linha["Alerta"] != "Sistema normal":
        print(f"{linha['Módulo']}: {linha['Alerta']}")

print("\nSistema finalizado.")
```

```python
historico=[]

def inserir_dados():
    temperatura=float(input("Temperatura da nave: "))
    energia=float(input("Nivel de energia (%): "))
    comunicacao=int(input("Comunicacao (1 = ativa / 0 = falha): "))
    leitura={
        "temperatura":temperatura,
        "energia":energia,
        "comunicacao":comunicacao
    }
    historico.append(leitura)
    print("Dados cadastrados com sucesso.")

def visualizar_status():
    if len(historico)==0:
        print("Nenhum dado cadastrado.")
    else:
        ultimo=historico[-1]
        print("Status atual da missao:")
        print("Temperatura:",ultimo["temperatura"],"C")
        print("Energia:",ultimo["energia"],"%")
        if ultimo["comunicacao"]==1:
            print("Comunicacao: ativa")
        else:
            print("Comunicacao: falha")

def executar_analise():
    if len(historico)==0:
        print("Nenhum dado para analisar.")
    else:
        ultimo=historico[-1]
        print("Analise operacional:")
        if ultimo["temperatura"]>80:
            print("Alerta de superaquecimento")
        elif ultimo["energia"]<20:
            print("Economia de energia")
        elif ultimo["comunicacao"]==0:
            print("Falha de comunicacao")
        else:
            print("Missao em estado normal")

def mostrar_historico():
    if len(historico)==0:
        print("Historico vazio.")
    else:
        print("Historico das leituras:")
        for i,dado in enumerate(historico):
            print("Leitura",i+1)
            print("Temperatura:",dado["temperatura"],"C")
            print("Energia:",dado["energia"],"%")
            print("Comunicacao:",dado["comunicacao"])

def menu():
    while True:
        print("\nSistema de Monitoramento Espacial")
        print("1 - Inserir dados")
        print("2 - Visualizar status")
        print("3 - Executar analise")
        print("4 - Historico das leituras")
        print("5 - Encerrar sistema")
        opcao=input("Escolha uma opcao: ")

        if opcao=="1":
            inserir_dados()
        elif opcao=="2":
            visualizar_status()
        elif opcao=="3":
            executar_analise()
        elif opcao=="4":
            mostrar_historico()
        elif opcao=="5":
            print("Sistema encerrado.")
            break
        else:
            print("Opcao invalida.")

menu()
```
